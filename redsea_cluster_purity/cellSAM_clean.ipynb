```python
import tifffile as tiff
import numpy as np
from skimage.morphology import remove_small_objects
from tqdm import tqdm

mask_path = "/mnt/jwh83-data/Confetti/output/intestine_data_mask_outputs/CellSAM_3_20260113_TIF/B004-A-404_rgb_cellSAM.tif"
out_path = "/mnt/jwh83-data/Confetti/output/intestine_data_mask_outputs/CellSAM_3_20260113_TIF/B004-A-404_cellSAM_label_clean_relabel.tif"

print("Loading mask...")
mask = tiff.imread(mask_path).astype(np.int32)

print("Removing small objects...")
mask_clean = remove_small_objects(mask, min_size=10)

print("Collecting labels...")
labels = np.unique(mask_clean)
labels = labels[labels != 0]

print(f"Relabeling {len(labels)} objects...")
mask_relabel = np.zeros_like(mask_clean, dtype=np.int32)

for new_id, lab in tqdm(
    enumerate(labels, start=1),
    total=len(labels),
    desc="Relabeling"
):
    mask_relabel[mask_clean == lab] = new_id

print("Saving output...")
tiff.imwrite(out_path, mask_relabel)

print(f"Saved: {out_path}")
print(f"Cells: {len(labels)}")
```
