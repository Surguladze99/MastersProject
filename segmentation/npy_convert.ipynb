```python
#!/usr/bin/env python3
from __future__ import annotations

from pathlib import Path
import numpy as np

try:
    import tifffile as tiff
except ImportError as e:
    raise SystemExit("Install dependency: pip install tifffile") from e


IN_DIR = Path("/mnt/jwh83-data/Confetti/output/intestine_data_mask_outputs/CellposeSAM")
OUT_DIR = Path("/mnt/jwh83-data/Confetti/output/intestine_data_mask_outputs/CellposeSAM_TIF")


def scale_array(arr: np.ndarray, mode: str) -> np.ndarray:
    """Min/max scale numeric array to uint8 or uint16 (useful for float data)."""
    if mode not in ("uint8", "uint16"):
        raise ValueError("mode must be 'uint8' or 'uint16'")

    a = arr.astype(np.float32, copy=False)
    finite = np.isfinite(a)
    if not finite.any():
        return np.zeros(arr.shape, dtype=np.uint8 if mode == "uint8" else np.uint16)

    mn = a[finite].min()
    mx = a[finite].max()
    if mx == mn:
        norm = np.zeros_like(a, dtype=np.float32)
    else:
        norm = (a - mn) / (mx - mn)
        norm = np.clip(norm, 0.0, 1.0)

    if mode == "uint8":
        return (norm * 255.0 + 0.5).astype(np.uint8)
    else:
        return (norm * 65535.0 + 0.5).astype(np.uint16)


def npy_to_tif_dir(in_dir: Path, out_dir: Path, scale: str | None = None) -> None:
    npy_files = list(in_dir.rglob("*.npy"))
    if not npy_files:
        print(f"No .npy files found under: {in_dir}")
        return

    converted = 0
    skipped = 0

    for npy_path in npy_files:
        rel = npy_path.relative_to(in_dir)
        tif_path = (out_dir / rel).with_suffix(".tif")
        tif_path.parent.mkdir(parents=True, exist_ok=True)

        try:
            data = np.load(npy_path, allow_pickle=False)
        except ValueError:
            # Some pipelines save object arrays; if that's your case, you can flip allow_pickle=True.
            print(f"SKIP (requires allow_pickle=True): {npy_path}")
            skipped += 1
            continue

        if not isinstance(data, np.ndarray):
            print(f"SKIP (not an ndarray): {npy_path}")
            skipped += 1
            continue

        if scale is not None:
            data = scale_array(data, scale)

        # Write (2D -> single image; 3D -> stack; higher dims also supported by tifffile)
        tiff.imwrite(str(tif_path), data)
        print(f"OK: {npy_path} -> {tif_path}")
        converted += 1

    print(f"\nDone. Converted: {converted}, Skipped: {skipped}")
    print(f"Output folder: {out_dir}")


if __name__ == "__main__":
    # Set scale to None to preserve dtype, or 'uint16'/'uint8' to rescale floats.
    npy_to_tif_dir(IN_DIR, OUT_DIR, scale=None)
```

```python
#!/usr/bin/env python3
from __future__ import annotations

from pathlib import Path
import numpy as np

try:
    import tifffile as tiff
except ImportError as e:
    raise SystemExit("Install dependency: pip install tifffile") from e


IN_DIR = Path("/mnt/jwh83-data/Confetti/output/intestine_data_mask_outputs/InstanSeg")
OUT_DIR = Path("/mnt/jwh83-data/Confetti/output/intestine_data_mask_outputs/20260212_InstanSeg_TIF")


def scale_array(arr: np.ndarray, mode: str) -> np.ndarray:
    """Min/max scale numeric array to uint8 or uint16 (useful for float intensity data)."""
    if mode not in ("uint8", "uint16"):
        raise ValueError("mode must be 'uint8' or 'uint16'")

    a = arr.astype(np.float32, copy=False)
    finite = np.isfinite(a)
    if not finite.any():
        return np.zeros(arr.shape, dtype=np.uint8 if mode == "uint8" else np.uint16)

    mn = a[finite].min()
    mx = a[finite].max()
    if mx == mn:
        norm = np.zeros_like(a, dtype=np.float32)
    else:
        norm = (a - mn) / (mx - mn)
        norm = np.clip(norm, 0.0, 1.0)

    if mode == "uint8":
        return (norm * 255.0 + 0.5).astype(np.uint8)
    else:
        return (norm * 65535.0 + 0.5).astype(np.uint16)


def _float_labels_to_int(data: np.ndarray) -> np.ndarray:
    """
    If a label mask is stored as float, convert it to an integer label image.

    - Rounds to nearest int to handle tiny float noise.
    - Forces background to 0 (if any values <= 0).
    - Uses uint32 by default to accommodate many labels.
    """
    a = np.asarray(data)
    a = np.nan_to_num(a, nan=0.0, posinf=0.0, neginf=0.0)

    # Round to nearest integer labels
    a = np.rint(a)

    # Ensure background is 0; clamp negatives to 0
    a[a < 0] = 0

    # If it fits in uint16, use that; otherwise uint32
    mx = float(a.max()) if a.size else 0.0
    if mx <= np.iinfo(np.uint16).max:
        return a.astype(np.uint16)
    return a.astype(np.uint32)


def npy_to_tif_dir(
    in_dir: Path,
    out_dir: Path,
    scale: str | None = None,
    assume_labels: bool = True,
) -> None:
    """
    Convert all .npy files under in_dir to .tif files under out_dir.

    If assume_labels=True and scale is None:
      - If the array is floating, it will be converted to an integer label image
        (rounded + clamped) so tools like skimage.regionprops / redsea work.

    If scale is 'uint8' or 'uint16':
      - Min/max scaling is applied (intended for intensity images, NOT label masks).
    """
    npy_files = list(in_dir.rglob("*.npy"))
    if not npy_files:
        print(f"No .npy files found under: {in_dir}")
        return

    converted = 0
    skipped = 0

    for npy_path in npy_files:
        rel = npy_path.relative_to(in_dir)
        tif_path = (out_dir / rel).with_suffix(".tif")
        tif_path.parent.mkdir(parents=True, exist_ok=True)

        try:
            data = np.load(npy_path, allow_pickle=False)
        except ValueError:
            print(f"SKIP (requires allow_pickle=True): {npy_path}")
            skipped += 1
            continue

        if not isinstance(data, np.ndarray):
            print(f"SKIP (not an ndarray): {npy_path}")
            skipped += 1
            continue

        # If user requested scaling, do it (for intensity images)
        if scale is not None:
            data = scale_array(data, scale)
        else:
            # If these are label masks, enforce integer dtype for compatibility
            if assume_labels and np.issubdtype(data.dtype, np.floating):
                data = _float_labels_to_int(data)

        # Write (2D -> single image; 3D -> stack; higher dims also supported by tifffile)
        tiff.imwrite(str(tif_path), data)
        print(f"OK: {npy_path} -> {tif_path}  (shape={data.shape}, dtype={data.dtype})")
        converted += 1

    print(f"\nDone. Converted: {converted}, Skipped: {skipped}")
    print(f"Output folder: {out_dir}")


if __name__ == "__main__":
    # For cell masks / label images:
    npy_to_tif_dir(IN_DIR, OUT_DIR, scale=None, assume_labels=True)

    # If you ever need intensity-image scaling instead, call:
    # npy_to_tif_dir(IN_DIR, OUT_DIR, scale="uint16", assume_labels=False)
```
