```python
import numpy as np
import pandas as pd
import tifffile as tiff


def count_rows_npy(npy_path):
    arr = np.load(npy_path)
    return arr.shape[0]


def count_rows_tif(tif_path):
    img = tiff.imread(tif_path)
    return img.shape[0]


def count_rows_csv(csv_path):
    df = pd.read_csv(csv_path)
    return len(df)


if __name__ == "__main__":
    npy_file = "/mnt/jwh83-data/Confetti/output/intestine_data_mask_outputs/CellposeSAM/B004-A-204_rgb_cellposeSAM.npy"
    tif_file = "/mnt/jwh83-data/Confetti/output/intestine_data_mask_outputs/CellposeSAM_TIF/B004-A-204_rgb_cellposeSAM.tif"
    csv_file = "/mnt/jwh83-data/Confetti/output/Redsea/CellposeSAM/B004-A-204/single_cell_after_redsea.csv"

    npy_rows = count_rows_npy(npy_file)
    tif_rows = count_rows_tif(tif_file)
    csv_rows = count_rows_csv(csv_file)

    print(f"NPY row count: {npy_rows}")
    print(f"TIF row count: {tif_rows}")
    print(f"CSV row count: {csv_rows}")
```

```python
import numpy as np
import tifffile as tiff


def count_cells_from_array(arr):
    """
    Counts cells in a labeled segmentation mask.
    Assumes:
      - background = 0
      - each cell has a unique integer label > 0
    """
    labels = np.unique(arr)
    labels = labels[labels != 0]  # remove background
    return len(labels)


def count_cells_npy(npy_path):
    arr = np.load(npy_path)
    return count_cells_from_array(arr)


def count_cells_tif(tif_path):
    arr = tiff.imread(tif_path)
    return count_cells_from_array(arr)


if __name__ == "__main__":
    npy_file = "/mnt/jwh83-data/Confetti/output/intestine_data_mask_outputs/CellposeSAM/B004-A-304_rgb_cellposeSAM.npy"
    tif_file = "/mnt/jwh83-data/Confetti/output/intestine_data_mask_outputs/CellposeSAM_TIF/B004-A-304_rgb_cellposeSAM.tif"

    npy_cells = count_cells_npy(npy_file)
    tif_cells = count_cells_tif(tif_file)

    print(f"NPY cell count: {npy_cells}")
    print(f"TIF cell count: {tif_cells}")
```
